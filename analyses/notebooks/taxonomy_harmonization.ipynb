# Taxonomic match proportions

Read completed backend taxonomy results without rerunning matching.
Image panels count every image once, including unclassified images.
Taxon panels count distinct non-null input-taxon keys referenced by the collection.
These are unique input taxa, not accepted species; several input taxa may map to one species.
Images without an input-taxon key cannot contribute to taxon-level denominators.
Both status and method proportions use their full respective population, including unresolved outcomes.
See [definitions](../README.md).


```python
import sys
from pathlib import Path

ROOT = next(
    p
    for p in (Path.cwd(), *Path.cwd().parents)
    if (p / "backend/app/configs/config.yaml").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
import matplotlib.pyplot as plt
from analyses.publication import bar_plot, export_figure, load_settings, publication_style

settings = load_settings(ROOT)
publication_style()
```

```python
from analyses.publication import taxonomy_summaries

summaries = taxonomy_summaries(settings)
```

## Status and method by counting unit


```python
fig, axes = plt.subplots(2, 2, figsize=(18, 12), layout="constrained")
for ax, key, title in zip(
    axes.flat,
    ("images_status", "taxa_status", "images_method", "taxa_method"),
    (
        "A  Match status — images",
        "B  Match status — unique input taxa",
        "C  Match method — images",
        "D  Match method — unique input taxa",
    ),
):
    bar_plot(ax, summaries[key], title)
export_figure(fig, settings, "taxonomy_match_proportions", summaries)
plt.show()
plt.close(fig)
```
