# Geographic metadata and coordinate validation

Coordinate availability requires both latitude and longitude to be finite numbers; it does not imply geographic validity.
Detailed locality means nonblank locality or verbatim-locality text, not country or state alone.
Validation categories come from completed geoharmonize output. Missing results are “Not evaluated.”
All proportions use all image records. No matching or database writes run here.
See [prerequisites](../README.md).


```python
import sys
from pathlib import Path

ROOT = next(
    p
    for p in (Path.cwd(), *Path.cwd().parents)
    if (p / "backend/app/configs/config.yaml").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
import matplotlib.pyplot as plt
from analyses.publication import (
    category_plot,
    export_figure,
    load_settings,
    publication_style,
)

settings = load_settings(ROOT)
publication_style()
```

```python
from analyses.publication import geography_summaries

summaries = geography_summaries(settings)
```

## Coordinate and locality availability


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5.5), layout="constrained")
category_plot(axes[0], summaries["coordinate_availability"], "A  Coordinate availability")
category_plot(axes[1], summaries["locality_availability"], "B  Detailed locality availability")
export_figure(
    fig,
    settings,
    "geographic_availability",
    {key: summaries[key] for key in ("coordinate_availability", "locality_availability")},
)
plt.show()
plt.close(fig)
```

## Coordinate-validation outcomes


```python
fig, ax = plt.subplots(figsize=(12, 6), layout="constrained")
category_plot(ax, summaries["coordinate_validation"], "Coordinate-validation outcomes")
export_figure(
    fig, settings, "coordinate_validation", {"outcomes": summaries["coordinate_validation"]}
)
plt.show()
plt.close(fig)
```
